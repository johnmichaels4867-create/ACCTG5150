```python
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf
from statsmodels.iolib.summary2 import summary_col
from sklearn.metrics import mean_squared_error, mean_absolute_error, mean_absolute_percentage_error

df = pd.read_csv('production_cost_daily.csv', parse_dates=['date'])
print(df.shape)
df.head(2)
```

```python
model_volume = smf.ols('total_cost ~ volume', data=df).fit()
print(model_volume.summary())
```

```python
from sklearn.metrics import mean_squared_error

df['total_cost_hat_e1'] = model_volume.predict(df)
rmse_e1 = np.sqrt(mean_squared_error(df['total_cost'], df['total_cost_hat_e1']))
print(f"RMSE: {rmse_e1:,.2f}")
```

```python
from sklearn.metrics import mean_absolute_error
mae_e1 = mean_absolute_error(df['total_cost'], df['total_cost_hat_e1'])
print(f"MAE: {mae_e1:,.2f}")
```

```python
from sklearn.metrics import mean_absolute_percentage_error
mape_e1 = mean_absolute_percentage_error(df['total_cost'], df['total_cost_hat_e1']) * 100
print(f"MAPE: {mape_e1:.2f}%")
```

```python
model_abc = smf.ols(
    'total_cost ~ material_lbs + labor_hours + machine_hours + num_batches + num_inspections',
    data=df
).fit()
print(model_abc.summary())
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, mean_absolute_percentage_error

df['total_cost_hat_e2'] = model_abc.predict(df)
print("RMSE:", np.sqrt(mean_squared_error(df['total_cost'], df['total_cost_hat_e2'])))
print("MAE: ", mean_absolute_error(df['total_cost'], df['total_cost_hat_e2']))
print("MAPE:", mean_absolute_percentage_error(df['total_cost'], df['total_cost_hat_e2']) * 100)
```

```python
model_qfe = smf.ols('total_cost ~ volume + C(quarter)', data=df).fit()
print(model_qfe.summary())
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, mean_absolute_percentage_error

df['total_cost_hat_e3'] = model_qfe.predict(df)
print("RMSE:", np.sqrt(mean_squared_error(df['total_cost'], df['total_cost_hat_e3'])))
print("MAE: ", mean_absolute_error(df['total_cost'], df['total_cost_hat_e3']))
print("MAPE:", mean_absolute_percentage_error(df['total_cost'], df['total_cost_hat_e3']) * 100)
```
