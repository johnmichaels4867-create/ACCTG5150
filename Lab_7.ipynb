```python
import pandas as pd

df = pd.read_excel("lab7_data.xlsx")

df['Invoice_Number'] = df['notes'].str.extract(r"Inv\w*\s*#:\s*([^\n]+)", expand=False)
df['Contact_Name']   = df['notes'].str.extract(r"Rep\w*:\s*([^\n]+)", expand=False)
df['Follow_Up_Date'] = df['notes'].str.extract(r"Follow[^:\n]*:\s*([^\n]+)", expand=False)
df['Clean_Notes']    = df['notes'].str.extract(r"Notes?:\s*([\s\S]+)", expand=False)

df.to_csv("lab7_data_cleaned.csv", index=False)

from google.colab import files
files.download("lab7_data_cleaned.csv")
```
